```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, classification_report
```

```python
dig = load_digits()

# Create dataframe
df = pd.DataFrame(dig.data, columns=dig.feature_names)

# Features and target
x = df
y = dig.target
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
model = SVC()

model.fit(x_train, y_train)
```

```python
accuracy = model.score(x_test, y_test)

print("Accuracy :", accuracy)

```

```python
y_pred = model.predict(x_test)
```

```python
plt.figure(figsize=(10,4))

for i in range(5):

    plt.subplot(1,5,i+1)

    # reshape 64 values into 8x8 image
    plt.imshow(dig.images[i], cmap='gray')

    plt.title(f"Label : {dig.target[i]}")

    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))

for i in range(10):

    plt.subplot(2,5,i+1)

    # x_test.iloc[i] contains 64 pixel values
    plt.imshow(x_test.iloc[i].values.reshape(8,8), cmap='gray')

    plt.title(f"Pred : {y_pred[i]}")

    plt.axis('off')

plt.tight_layout()
plt.show()

```

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(10,7))

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")

plt.show()
```

```python
print(classification_report(y_test, y_pred))
```
