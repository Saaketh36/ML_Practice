```python

import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("train.csv")

print(df.info())
```

```python
df = df[[
    'SalePrice',
    'LotArea',
    'OverallQual',
    'YearBuilt',
    'GarageCars',
    'Neighborhood'
]]
```

```python

df.isnull().sum()

df['GarageCars'] = df['GarageCars'].fillna(
    df['GarageCars'].mean()
)

df['Neighborhood'] = df['Neighborhood'].fillna(
    'Unknown'
)
```

```python
le = LabelEncoder()

df['Neighborhood'] = le.fit_transform(
    df['Neighborhood'])
```

```python
df['HouseAge'] = 2025 - df['YearBuilt']

scaler = StandardScaler()

df[['LotArea']] = scaler.fit_transform(
    df[['LotArea']]
)
```

```python
correlation = df['SalePrice'].corr(
    df['OverallQual']
)

print("Correlation :", correlation)
print(df.corr(numeric_only=True))
```

```python
covariance = df['SalePrice'].cov(
    df['OverallQual']
)

print("Covariance :", covariance)
```
